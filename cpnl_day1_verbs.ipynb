## Goal: get a version of the deck that teaches the present indicative of the 12 verbs that CPNL wants you to memorize first. All others will be in the same order underneath

```python
import pandas as pd
```

```python
verbs = ['ser', 'estar', 'tenir', 'voler', 'poder', 'fer',
         'anar', 'venir', 'tornar', 'donar', 'dir',' saber']

df = pd.read_csv('table_to_make_cards.csv', sep='|')

# get inds of these verbs in present indicative
inds = df.loc[(df.inf_verb.isin(verbs))&\
       (df.tense == 'P')&\
       (df.mood == 'I')].index

print(len(df.index))

# copy table with just these verbs
pi_day1_verbs_df = df.loc[inds].copy(deep=True)

print(len(pi_day1_verbs_df.index))

# remove the entries from original df
df = df.drop(index=inds)

print(len(df.index))
```

```python
# reconcat 
df = pd.concat([pi_day1_verbs_df, df], axis=0)
print(len(df.index))
```

```python
df.to_csv('table_to_make_cards_cpnl_day1_verbs.csv', sep='|')
```
